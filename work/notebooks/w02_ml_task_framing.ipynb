## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*A fixed rule like "stale (180+ days unchanged) AND visible (500+ impressions)"
only flagged 17 out of 30,000 pages — far too narrow to be a usable review
queue. It also can't weigh signals against each other: it treats staleness
and visibility as strict yes/no gates, when in reality a page with 400
impressions and 150 days since update might carry more real signal than
one that barely clears both hard thresholds. A learned model can combine
impressions, position, freshness, and engagement as continuous, weighted
signals instead of hard cutoffs — which is exactly why it triples the
rule's Precision@50 (0.240 → 0.740) despite using the same underlying data.
