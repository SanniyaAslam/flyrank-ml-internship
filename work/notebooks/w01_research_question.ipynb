## 2. The question: decision, action, cost of a wrong call

Decision: Which content pages should a content reviewer prioritize for refresh, given limited review capacity.
Who acts: A content/SEO reviewer or strategist working through a queue of pages.
Action: Review the top-ranked pages first and decide whether to refresh, expand, protect, or leave them.
Cost of a wrong call: If the model flags a page as high-priority but it isn't actually declining, the reviewer wastes limited time on a low-value page while a genuinely declining page goes unreviewed — a false positive costs reviewer hours; a false negative costs lost visibility that goes unnoticed.

## 3. Quick look at the data (2-3 real numbers)

Loading the starter dataset (data/raw/content_refresh_anonymized.csv) confirms this lane has real signal to work with:

30,000 rows — a solid starter slice to build and validate a baseline model on.
54.2% of pages are labeled "declining" (trend_direction == "down") — a large enough positive class that a model has plenty of examples to learn from, without being so rare that classification becomes unreliable.
Declining pages average 4,919 impressions and 35 sessions over 90 days, versus 5,533 impressions and 40 sessions for non-declining pages — a real but modest gap, suggesting decline isn't only about low-traffic pages disappearing quietly; even fairly visible pages show measurable decline.
Only 17 of 30,000 pages meet the "stale visible page" reason-code (unchanged 180+ days and 500+ impressions) — showing that a naive stale-content rule alone would flag almost nothing, which is exactly the gap Lane 2's ranked scoring is meant to close.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total rows:", len(df))
print("Declining %:", (df["trend_direction"] == "down").mean() * 100)

print(df.groupby(df["trend_direction"] == "down")[["impressions_90d", "sessions_90d"]].mean())

stale = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]
print("Stale visible pages:", len(stale), "out of", len(df))
```

What this work can say: My scoring will identify pages that show observed patterns of decline based on measured signals — impressions, position, freshness, engagement. The starter numbers already show these signals carry real information (e.g. a 600-impression average gap between declining and non-declining pages). The output is decision-support: a ranked queue that helps a reviewer prioritize limited review time, with reason codes explaining why each page was flagged. Any pattern I find between signals and decline is directional and observational — I watched what happened in the data, I did not run an experiment.

What this work can never say: This project cannot prove that refreshing a page causes it to recover — that would need a controlled experiment (e.g. comparing refreshed vs. untouched pages over time), which this dataset doesn't support. It also cannot claim to have discovered or predicted anything about Google's actual ranking algorithm — I'm only measuring observed search/engagement outcomes, never the algorithm itself.
